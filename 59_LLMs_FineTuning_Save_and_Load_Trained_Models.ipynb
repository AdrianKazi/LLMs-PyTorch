# Intro

* How to save (to disk) and download fined-tuned models, and how to import them back into Python
* The difference between saving/loading a model via HuggingFace model methods vs. the PyTorch save function

# Import

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

tokenizer = AutoTokenizer.from_pretrained("gpt2")
gpt2 = AutoModelForCausalLM.from_pretrained("gpt2")
```

# Save

```python
gpt2.save_pretrained('gpt_original_hf')
```

# Replace Embeddings with 1's

```python
gpt2.transformer.wte.weight
```

```python
gpt2
```

```python
gpt2.transformer.wte.weight.data = torch.ones_like(gpt2.transformer.wte.weight.data)
gpt2.transformer.wte.weight.data
```

```python
gpt2.save_pretrained('gpt_with_ones_hf')
```

# Clear the model from the workspace

```python
del gpt2
```

# Import 2 modles

```python
from transformers import GPT2LMHeadModel

orig_model = GPT2LMHeadModel.from_pretrained('gpt_original_hf')
ones_model = GPT2LMHeadModel.from_pretrained('gpt_with_ones_hf')

orig_model, ones_model
```

# Savind and loading the model state (non-hf mdoels)

```python
torch.save(orig_model.state_dict(), 'gpt_original_state_dict.pt')
```

Load it back -- replacing the model with all 1's



```python
ones_model.load_state_dict(torch.load('gpt_original_state_dict.pt'))
ones_model.transformer.wte.weight
```
