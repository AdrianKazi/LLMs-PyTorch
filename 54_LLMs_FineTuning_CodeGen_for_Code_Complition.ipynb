```python
import torch
import torch.nn as nn
import numpy as np

# GPU
device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
device
```

# Import and Inspect the CodeGen Model

```python
# Model source:
# https://huggingface.co/Salesforce/codegen-350M-mono
from transformers import AutoTokenizer, AutoModelForCausalLM
tokenizer = AutoTokenizer.from_pretrained('Salesforce/codegen-350M-mono')

model = AutoModelForCausalLM.from_pretrained(
    "Salesforce/codegen-350M-mono",
    torch_dtype=torch.float32
).to(device)
model
```

# Check out the Tokenizer

```python
 # A bit about their tokenizer
 print(f'Tokenizer has {tokenizer.vocab_size:,} tokens.\n')

 for i in range(30):
  # generate random token
  rand_tok = torch.randint(tokenizer.vocab_size, (1,))
  print(f'Token {rand_tok[0]:5} is "{tokenizer.decode(rand_tok)}"')
```

# Generate a response to some code input

```python
text = 'for i in range(10):'
input_ids = tokenizer(text, return_tensors='pt').input_ids.to(device)

generated_ids = model.generate(input_ids, max_length = 222, temperature = 1.4, do_sample = True)
print(tokenizer.decode(generated_ids[0], skip_special_tokens = True))
```

```python
import os
import subprocess
import nbformat

# download the repo
repo_url = 'https://github.com/mikexcohen/Calculus_book.git'
repo_dir = 'Calculus_book'
subprocess.run(['git', 'clone', repo_url, repo_dir])

# initialize a list for all tokens
all_tokens = []

## find all .ipynb files
for root, dirs, files in os.walk(repo_dir):
  for file in files:
    if file.endswith('.ipynb'):

      # load and process noetbooks
      # print(f'Processing notebook: {os.path.join(root, file)}')
      with open(os.path.join(root, file), 'r', encoding = 'utf-8') as f:

        # parse the notebook structure
        nb_data = nbformat.read(f, as_version=4)

        # gather all code cells into a single string
        code_cells = []
        for cell in nb_data['cells']:
          if cell['cell_type'] == 'code': # this cell contains code
            code_cells.append(cell['source']) # the actual code in this cell

        # join code cells
        all_code = '\n'.join(code_cells)

        # tokenize and add to the list of all tokens
        all_tokens += tokenizer(all_code)['input_ids']

# check the final dataset size
print(f'\n\nTraining data contains {len(all_tokens):,} tokens, of which {len(set(all_tokens))} are unique.')
```

* Always look for specialized pretrained models! They often come in multiple sizes (parameter counts)
* You have now seen how code-completion AI tools work

```python
# need the tokens to be a tensor type
all_tokens = torch.tensor(all_tokens, dtype = torch.long)
```

# Fine-Tune the Model

```python
# optimizer
optimizer = torch.optim.AdamW(model.parameters(), lr = 5e-5)

# training parameters
seq_len     = 128
batch_size  = 16
num_samples = 200
```

```python
# traniing loop

# initialize losses
losses = np.zeros(num_samples)

for sampli in range(num_samples):

  # get batch of data
  ix = torch.randint(len(all_tokens)-seq_len,size=(batch_size,))
  X  = all_tokens[ix[:,None] + torch.arange(seq_len)].to(device)

  # forward pass and get loss
  model.zero_grad()
  outputs = model(X, labels=X)

  # backprop and store loss
  outputs.loss.backward()
  optimizer.step()
  losses[sampli] = outputs.loss.item()

  # update progress display
  if sampli%42 == 0:
    print(f'Sample {sampli:4}/{num_samples}, losses: {losses[sampli]:.4f}')
```

# Evaluate

```python
# pretrained GPTs are better at context-informed token generation
text = 'for i in range(10):'
input_ids = tokenizer(text, return_tensors='pt').input_ids.to(device)

generated_ids = model.generate(input_ids, max_length=250, temperature=1.4, do_sample=True)
print(tokenizer.decode(generated_ids[0], skip_special_tokens=True))
```
