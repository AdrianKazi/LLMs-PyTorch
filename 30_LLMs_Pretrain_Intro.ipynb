# Pretrain vs fine-tune vs instruction tune

| | Pretraining | Fine-Tuning | Instruction tuning |
|-|-------------|-------------|-------------|
| **Purpose**| Learn general language structure | Adapt the model to a specified task | Align model with instructions |
|**Data**| Huge (most of the web) | Curated and domain-specific | Human-written prompts|
|**Training**|Unsupervised;very long and expensive|Shorter (hours or days)|Next-token prediction and RLHF|
|Applications|Base model. Mechanistic interpret.|Task-specific applications|Chatbots, assistants, general purpose|
