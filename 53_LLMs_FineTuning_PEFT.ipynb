# Intro

* What PEFT means
* How the PEFT family of methods can help train large models on limited resources
* About some resources for learning more about implementing PEFT

* PEFT: A family of techniques for freezing and fine-tuning pretrained LLMs
* General Idea: Freeze as much of the model as possible, possibly adding in new parameters at key locations.
* Advantage: Train an LLM with limited resources
* Disadvantages: Difficult to match performance, generalization to new tasks may be limited

PEFT Examples

* Adapters: Freeze the entire model, add small (trainable) layers at key model locations

* LoRA / DoRA: Decompose weights matrices to low-rank approximations, and train those low-rank matrices

* Prefix module: Train a "pre-embeddings" layer

* BitFit: Train only the bias terms (not the weights)

* PEFT is a set of related techniques that allow you to fine-tune models with limited computational resources
* PEFT methods are very restrictive, and tend to work well when the task is well-characterized, e.g., classification
* Running and fine-tuning very large models still requires considerable resources
